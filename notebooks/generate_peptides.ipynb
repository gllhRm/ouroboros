# Generate a peptide batch

Choose the number of peptides, their length bounds, and a seed. Equal bounds give every peptide the same length. Run the cells from top to bottom to preview the sequences and export them as FASTA.

```python
import sys
from pathlib import Path

from ouroboros import generate_peptides, save_fasta, write_fasta

count = 2
min_length = 8
max_length = 8
seed = 42
```

## Preview the FASTA records

Each length bound is inclusive and can range from 2 to 50 amino acids. The settings above reproduce the two records in the README.

```python
peptides = list(generate_peptides(count, min_length, max_length, seed=seed))
write_fasta(peptides, sys.stdout)
```

## Save the batch

The next cell saves the displayed sequences to `example_peptides.fasta` in the kernel's working directory. Change the filename to keep multiple batches. Reusing the same filename replaces the previous export.

```python
output_path = Path("example_peptides.fasta")
save_fasta(peptides, output_path)
print(f"Saved {len(peptides)} peptides to {output_path.name}")
```

# Inspect peptide lengths

Count the peptides at each length in a seeded random batch. This notebook generates its own sample. Change the seed or batch size and rerun the cells to compare samples.

```python
from collections import Counter

from ouroboros import generate_peptides

count = 200
min_length = 5
max_length = 12
seed = 7
```

## Count peptides by length

Each length in the selected range has the same chance of being sampled. The table includes lengths observed zero times and the total number of sampled peptides.

```python
batch = generate_peptides(count, min_length, max_length, seed=seed)
length_counts = Counter(len(peptide) for peptide in batch)

print("Length (aa)  Peptides")
for length in range(min_length, max_length + 1):
    print(f"{length:>11}  {length_counts[length]:>8}")
print(f"{'Total':>11}  {sum(length_counts.values()):>8}")
```
